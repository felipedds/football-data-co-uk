```sql
%python
import pandas as pd
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("premier_league_gold").getOrCreate()

# File
consolidated_data_gold = "/Workspace/Repos/felipediasd@gmail.com/football-data-co-uk/data/gold/premier_league/enriched_data.parquet"

# Read o CSV
df_gold = spark.createDataFrame(pd.read_parquet(consolidated_data_gold))

# Create a view
df_gold.createOrReplaceTempView("gold_table")
```

```sql
%sql
SELECT 
HomeTeam, COUNT(*) as total_matches
FROM gold_table
GROUP BY HomeTeam
ORDER BY total_matches DESC
```
