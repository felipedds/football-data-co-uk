Import libraries

```python
import pyspark
```

Delete Catalog and Schemas

```python
# Environments = Catalogs
environments = ["dev_football_data_uk"]

# Country → League mapping = Schema
leagues = {
    "england": "premier_league",
    "spain": "la_liga",
    "germany": "bundesliga",
    "italy": "serie_a",
    "france": "ligue_1"
}

# Medallion layers
layers = ["bronze", "silver", "gold"]

# Drop catalogs and schemas
for env in environments:
    # Drop catalog for environment
    spark.sql(f"DROP CATALOG IF EXISTS {env} CASCADE;")
    print(f"Dropped catalog: {env}")

    # Drop schemas
    for country, league in leagues.items():
        for layer in layers:
            schema_name = f"{env}.{country}_{league}_{layer}"
            spark.sql(f"DROP SCHEMA IF EXISTS {schema_name} CASCADE;")
            print(f"Dropped schema: {schema_name}")
```
