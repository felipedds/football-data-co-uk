Import libraries

```python
import pyspark
```

Create Catalog and Schemas

```python
# Environment (Catalog)
environments = ["dev"]

# Domain
domain = "football_data_uk"

# Medallion layers (Schemas)
layers = ["raw", "bronze", "silver", "gold"]

# Create catalogs and schemas
for env in environments:
    # Create catalog
    spark.sql(f"CREATE CATALOG IF NOT EXISTS {env};")
    
    for layer in layers:
        schema_name = f"{env}.{domain}_{layer}"
        spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema_name};")
        print(f"Created schema: {schema_name}")

        # Create volume only for RAW layer
        if layer == "raw":
            volume_name = f"{schema_name}.files"
            spark.sql(f"CREATE VOLUME IF NOT EXISTS {volume_name};")
            print(f"Created volume: {volume_name}")
```
