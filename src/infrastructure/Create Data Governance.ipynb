Import libraries

```python
import pyspark
```

Create Catalog and Schemas

```python
# Environments = Catalogs
environments = ["dev_football_data_uk"]

# Country → League mapping = Schema
leagues = {
    "england": "premier_league",
    "spain": "la_liga",
    "germany": "bundesliga",
    "italy": "serie_a",
    "france": "ligue_1"
}

# Medallion layers
layers = ["bronze", "silver", "gold"]

# Create catalogs and schemas
for env in environments:
    # Create catalog for environment
    spark.sql(f"CREATE CATALOG IF NOT EXISTS {env};")
    
    for country, league in leagues.items():
        for layer in layers:
            schema_name = f"{env}.{country}_{league}_{layer}"
            spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema_name};")
            print(f"Created schema: {schema_name}")
```
